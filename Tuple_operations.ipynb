```python
#Check if the number is divisible by 4 and 7.

num = int(input("Enter any number"))
if (num % 4 == 0) and (num % 7 == 0):
  print (num,"is a divisible by both 4 and 7")
else:
  print (num,"is not divisble by 4 and 7")
```
